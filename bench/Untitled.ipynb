```python
import time
```

```python
import polars as pl
```

```python
path = "/home/beck/GolandProjects/ursus/bench/data/h2o/n10000000/groupby/g1.csv"
```

```python
s = time.perf_counter()
lazy_df = pl.scan_csv(path)
# 2. Build the query
query = lazy_df.select(
    [
        # Total length of the dataset
        pl.len().alias("total_count"),
        # Specific aggregations requested
        pl.col("id4").sum().alias("id4_sum"),
        pl.col("id5").mean().alias("id5_mean"),
        # Sum of the remaining numeric columns
        pl.col("id6").sum().alias("id6_sum"),
        pl.col("v1").sum().alias("v1_sum"),
        pl.col("v2").sum().alias("v2_sum"),
        pl.col("v3").sum().alias("v3_sum"),
    ]
)

# 3. Execute and print results
result = query.collect()
print(result)
print(time.perf_counter() - s)
```
